```python
def convert_to_list(name):
    if isinstance(name, str):
        return [name.strip()]
    return []
```

```python
df['director'] = df['director'].apply(convert_to_list)
df['cast'] = df['cast'].apply(convert_to_list)
```

```python
def collapse(L):
    L1 = []
    for i in L:
        L1.append(i.replace(" ",""))
    return L1
```

```python
df['director'] = df['director'].apply(collapse)
df['genre'] = df['genre'].apply(collapse)
df['cast'] = df['cast'].apply(collapse)
```

```python
import nltk
from nltk.stem.porter import PorterStemmer
from sklearn.feature_extraction.text import CountVectorizer
```

```python
ps = PorterStemmer() #class ka object
ps.stem('abusing') # trial
ps.stem('abused')
ps.stem('abuse')
```

```python
def stem(text):
    y =[]

    for i in text.split():
        y.append(ps.stem(i))

    return " ".join(y) #will convert back into string
```

```python
print(stem('accept accepts accepting accepted'))
```

```python
dff['tags'][0]
```

```python
stem('action thriller a high-octane action thriller which outlines the emotional journey of a man who is set to rectify the wrongs in the society. atlee shahrukhkhan,nayanthara,vijaysethupathi,deepikapadukone')
```

```python
dff['tags'] = dff['tags'].apply(stem)
dff.sample(5)
```

```python
# now will do text vectorization
cv = CountVectorizer(max_features=5000,stop_words='english')
#using cv we will convert it into array
vector = cv.fit_transform(dff['tags']).toarray()
vector.shape
```

```python
def recommend(movie):
    movie = movie.strip().lower()
    matches = dff[dff['movie_name'].str.lower() == movie]

    if matches.empty:
        print("Movie not found!")
        return

    

    index = dff[dff['movie_name'].str.strip().str.lower() == movie].index[0]  #index fetching

    distances = sorted(list(enumerate(similarity[index])),reverse=True,key=lambda x: x[1]) #distance after enumerating

    for i in distances[1:6]:
        print(dff.iloc[i[0]].movie_name) #will print movie name not id
```

```python
recommend('Jawan')
```
